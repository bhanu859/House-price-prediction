```python
X=df[["Area_sqft","Bedrooms","Bathrooms","Age_years","Parking"]]
```

```python
y=df["Price_lakhs"]
```

```python
X
```

```python
X.shape
print("X.shape",X.shape)
```

```python
y.shape
print("y.shape",y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(
    X,
    y,
    test_size = 0.3,
    random_state=42
)
```

```python
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
print("X_train",X_train.shape)
print("X_test",X_test.shape)
print("y_train",y_train.shape)
print("y_train",y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
print("Intercept:",model.intercept_)
print("coefficient:",model.coef_)
```

```python
coefficient=pd.DataFrame({
    "feature":X.columns,
    "coefficient":model.coef_
})
```

```python
print(coefficient)
```

```python
y_pred= model.predict(X_test)
```

```python
print("Predicted prices:")
print(y_pred)
```

```python
comparision=pd.DataFrame({
    "Actual price": y_test.values,
    "Predicted price": y_pred
})
```

```python
print(comparision)
```

```python
comparision["error"] = comparision["Actual price"] - comparision["Predicted price"]
```

```python
print(comparision)
```

```python
comparision["Absolute error"]=abs(
    comparision["error"]
)
```

```python
print(comparision)
```

```python
from sklearn.metrics import (mean_absolute_error,mean_squared_error,r2_score)
import numpy as np
```

```python
mae=mean_absolute_error(y_test,y_pred)
print("MAE:",mae)
```

```python
mse=mean_squared_error(y_test,y_pred)
print("MSE:",mse)
```

```python
rmse=np.sqrt(mse)
print("RMSE:",rmse)
```

```python
from sklearn.metrics import r2_score as sklearn_r2_score

r2_value = sklearn_r2_score(y_test, y_pred)
print("r2_score:", r2_value)
```

```python
print("===== Model performance =====")
print(f"MAE:{mae:.2f} lakhs")
print(f"MSE:{mse:.2f}")
print(f"RMSE:{rmse:.2f} lakhs")
print(f"r2:{r2_value:.3f}")
```

```python
comparision=pd.DataFrame({
    "Actual price":y_test.values,
    "predicted price":y_pred
})
print(comparision.round(2))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.scatter(y_test,y_pred)
min_price=min(y_test.min(),y_pred.min())
max_price=max(y_test.max(),y_pred.max())
plt.plot(
    [min_price,max_price],
    [min_price,max_price]
)
plt.xlabel("Actual price")
plt.ylabel("predicted price")
plt.title("Actual vs predicted house prices")
plt.show()
```

```python
errors=y_test-y_pred
plt.figure(figsize=(8,5))
plt.scatter(y_pred,errors)
plt.axhline(y=0,linestyle='--')
plt.xlabel("predicted price")
plt.ylabel("residual")
plt.title("predicted errors")
plt.show()
```
